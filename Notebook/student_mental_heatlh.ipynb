```python
from sklearn.model_selection  import train_test_split
from sklearn.preprocessing import StandardScaler,OrdinalEncoder,OneHotEncoder,FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression,Lasso,Ridge,ElasticNet
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score,mean_absolute_error
from sklearn.svm import SVR
from sklearn.ensemble  import GradientBoostingRegressor
```

```python
x=df.drop('Mental_Health_Score',axis=1)
y=df['Mental_Health_Score']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
df.columns
```

```python
log_col=['Study_Hours']
num_cols=['Age',
'Avg_Daily_Usage_Hours',     
'Daily_Unlocks',              
'Study_Hours',                
'Physical_Activity_Hours',    
'Sleep_Hours_Per_Night']
ordinal_col=['Stress_Level']
cat_cols=['Gender','Country','Academic_Level','Most_Used_Platform','Purpose_Of_Use']
```

```python
print(type(log_col), log_col)
print(type(num_cols), num_cols)
print(type(cat_cols), cat_cols)
print(type(ordinal_col), ordinal_col)
```

```python
num_pip=Pipeline(([('impute',SimpleImputer()),('scale',StandardScaler())]))
ordinal_pip=Pipeline([('odrdinal',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))])
cat_pip=Pipeline([('ohe',OneHotEncoder(sparse_output=False,handle_unknown='ignore'))])
log_pip=Pipeline([('log_transform',FunctionTransformer(np.log1p)),
                  ('scale',StandardScaler())])

preprocessing=ColumnTransformer([('log_pip',log_pip,log_col),
                                 ('num_pip',num_pip,num_cols),
                                 ('cat_pip',cat_pip,cat_cols),
                                 ('ordi_pip',ordinal_pip,ordinal_col)]
                                 ,remainder='passthrough')
model_pip=Pipeline([('preprocessing',preprocessing),('algo',LinearRegression())])
```

```python
model_pip.fit(x_train,y_train)
```

```python
x_train.columns
```

```python
y_pred=model_pip.predict(x_test)
```

```python
print(r2_score(y_test,y_pred))
```

```python
model_pip_random=Pipeline([('preprocessing',preprocessing),('algo_R',RandomForestRegressor())])
model_pip_random.fit(x_train,y_train)
y_pred=model_pip_random.predict(x_test)
print(r2_score(y_test,y_pred))
```

```python
model_pip_T=Pipeline([('preprocessing',preprocessing),('algo_T',DecisionTreeRegressor())])
model_pip_T.fit(x_train,y_train)
y_pred=model_pip_T.predict(x_test)
print(r2_score(y_test,y_pred))
```

```python
model_pip_GB=Pipeline([('preprocessing',preprocessing),('algo_GB',GradientBoostingRegressor())])
model_pip_GB.fit(x_train,y_train)
y_pred=model_pip_GB.predict(x_test)
print(r2_score(y_test,y_pred))
```
